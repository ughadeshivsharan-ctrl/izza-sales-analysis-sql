```python
import pandas as pd
from sqlalchemy import create_engine

path = r"C:\Users\LENOVO\Documents\python project\New folder\basic python.py\pizza.analysis.py\all excel_file"

orders = pd.read_csv(path + r"\orders.xlsx.csv")
orders_details = pd.read_csv(path + r"\orders_details.xlsx.csv")
pizzas = pd.read_csv(path + r"\pizzas.xlsx.csv")
pizza_types = pd.read_csv(path + r"\pizza_types.xlsx.csv", encoding="latin1")

engine = create_engine(
    "mysql+pymysql://root:Shiva%403613@localhost:3306/pizzahut_analysis"
)

orders.to_sql("orders", engine, if_exists="replace", index=False)
orders_details.to_sql("orders_details", engine, if_exists="replace", index=False)
pizzas.to_sql("pizzas", engine, if_exists="replace", index=False)
pizza_types.to_sql("pizza_types", engine, if_exists="replace", index=False)

print("Data uploaded successfully")
```
